```python
import pandas as pd
import glob

# Path to your folder containing all CSV files
path = "../cleaned_data/*.csv"

all_files = glob.glob(path)
df_merged = pd.concat([pd.read_csv(f) for f in all_files], ignore_index=True)

print(f"Total rows after merging: {len(df_merged)}")
print(f"Total files merged: {len(all_files)}")

df_merged.to_csv("merged_data.csv", index=False)
print("Merged file saved successfully!")
```
